```python
import pandas as pd
import numpy as np
import math
import scipy.stats.wilcoxon as wilcoxon
```

```python
df = pd.DataFrame({'before':[1,5,6,8,9,10], 'after': [2,4,12,14,18,20]})
df
```

```python
# нулевая гипотеза: интенсивность сдвигов в типичном направлении не превосходит интенсивности сдивгов в нетипичном
# альтернативная гипотеза: интенсивность сдвигов в типичном направлении превосходит интенсивности сдивгов в нетипичном
```

```python
shifts = df['after'] - df['before']
df['shifts'] = shifts
df
```

```python
df['abs_shifts'] = df['shifts'].abs()
df
```

```python
df.sort_values(by = 'abs_shifts')
df
```

```python
sum(df['abs_shifts'])
sum_1 = sum(df[df['shifts'] < 0]['abs_shifts'])
sum_2 = sum(df[df['shifts'] > 0]['abs_shifts'])
print(sum_1, sum_2)
```

```python
df['abs_shifts'].rank(method = 'average')
```
